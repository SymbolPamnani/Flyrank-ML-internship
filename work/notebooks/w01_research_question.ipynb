## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

### 3. Answer

The starter dataset contains **30,000 content pages** across **44 original columns**. It includes search-performance measures such as impressions, clicks, sessions, average position, and recent-versus-previous 30-day performance.

Of the 30,000 pages, **26,612** have a measurable change in impressions between the previous and latest 30-day periods. Among these, **19,716 pages (74.1%)** show declining impressions. The average impression change across measurable pages is **-4.8%**.

These measurements suggest that declining search impressions are common in this dataset, making content-refresh prioritization a reasonable decision problem to investigate.


```python
import pandas as pd

DATA_PATH = "content_refresh_anonymized.csv"

df = pd.read_csv(DATA_PATH)

# Calculate impression change from previous 30 days to last 30 days
df["impression_change_pct"] = (
    (df["impressions_last_30d"] - df["impressions_prev_30d"])
    / df["impressions_prev_30d"].replace(0, pd.NA)
) * 100

# Basic dataset size
print(f"Total pages: {len(df):,}")
print(f"Original columns: {len(df.columns) - 1:,}")

# Pages with a measurable impression change
measurable = df["impression_change_pct"].dropna()

print(f"Pages with measurable impression change: {len(measurable):,}")

# Declining pages
declining = (measurable < 0).sum()
declining_pct = declining / len(measurable) * 100

print(f"Pages with declining impressions: {declining:,}")
print(f"Share with declining impressions: {declining_pct:.1f}%")

# Average impression change
print(f"Average impression change: {measurable.mean():.1f}%")
```
